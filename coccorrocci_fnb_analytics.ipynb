```python
from dataclasses import dataclass
import pandas as pd

@dataclass
class DailyMetrics:
  date: str
  coperti_ristorante: float
  ordini_asporto: int
  scontrino_asporto: float
  food_cost_effettivo: float
  scontrino_ristorante: float

class RestaurantAnalytics:
  def __init__(self, canone_fisso_annuo: float = 20000.0, giorni_stagione: int = 120):
    self.canone_giornaliero = canone_fisso_annuo / giorni_stagione

  def calculate_daily_revenue(self, metrics: DailyMetrics) -> dict:
    ricavo_ristorante = metrics.coperti_ristorante * metrics.scontrino_ristorante
    ricavo_asporto = metrics.ordini_asporto * metrics.scontrino_asporto
    ricavo_totale = ricavo_ristorante + ricavo_asporto

    incidenza_food_cost = (metrics.food_cost_effettivo / ricavo_totale * 100) if ricavo_totale > 0 else 0

    costo_personale = ricavo_totale * 0.28
    utente_altri = ricavo_totale * 0.065
    ebitda_giornaliero = ricavo_totale - (metrics.food_cost_effettivo + costo_personale + utente_altri + self.canone_giornaliero)

    return {
      "ricavo_totale": round(ricavo_totale, 2),
      "ricavo_ristorante": round(ricavo_ristorante, 2),
      "ricavo_asporto": round(ricavo_asporto, 2),
      "incidenza_food_cost_pct": round(incidenza_food_cost, 1),
      "ebitda_giornaliero": round(ebitda_giornaliero, 2)
    }

print("✅ Modulo Analytics caricato con successo!")
```

```python
from re import X
import numpy as np
from sklearn.ensemble import RandomForestRegressor

class StockPredicatorAi:
  def __init__(self):
    self.model = RandomForestRegressor(n_estimators=50, random_state=42)
    self._train_model()

  def _train_model(self):
    # Feature storico: [presenze_campeggio, temperatura_max, e_weekend( 0 o 1)]
    X_train = np.array([
        [150, 26, 0], [200, 28, 1], [300, 31, 1],
        [350, 33, 1], [380, 34, 0], [400, 35, 1],
        [180, 27, 0], [250, 30, 1], [100, 24, 0]
    ])
    # Target: Coperti effettivi serviti
    y_train = np.array([45, 65, 85, 95, 90, 110, 50, 5, 30])
    self.model.fit(X_train, y_train)

  def predict_coperti_and_supplies(self, presenze_campeggio: int, temp: float, is_weekend: bool) -> dict:
    X_test = np.array([[presenze_campeggio, temp, 1 if is_weekend else 0]])
    coperti_stimati = int(self.model.predict(X_test)[0])

    chili_carne_pesce = round(coperti_stimati * 0.25, 1)
    panetti_pizza = int(coperti_stimati * 0.4)

    return {
      "coperti_previsti": coperti_stimati,
      "fabbisogno_carne_pesce_kg": chili_carne_pesce,
      "panetti_pizza_stimati": panetti_pizza
    }

print("✅ Modulo AI Stock Predictor pronto!")
```

```python
# Test della simulazione
analytics = RestaurantAnalytics(canone_fisso_annuo=20000)
today_data = DailyMetrics(
    date = "2026-08-15",
    coperti_ristorante = 95,
    scontrino_ristorante = 26.50,
    ordini_asporto = 45,
    scontrino_asporto = 13.00,
    food_cost_effettivo = 850.00
)

res = analytics.calculate_daily_revenue(today_data)
print(f"📊 RISUTLATI GIORNALIERI COCCORROCCI ({today_data.date}):")
print(f"  - Ricavo totale: €{res['ricavo_totale']}")
print(f"  - Food Cost%: {res['incidenza_food_cost_pct']}%")
print(f"  - Ebitda Giornaliero Netto: €{res['ebitda_giornaliero']}\n")

ai = StockPredicatorAi()
pred = ai.predict_coperti_and_supplies(presenze_campeggio=380, temp=33.0, is_weekend=True)
print(f"🤖 PREVISIONE AI PER DOMANI (Campeggio: 380 presenze, Ferragosto Weekend):")
print(f"  - Coperti Stimati: {pred['coperti_previsti']}")
print(f"  - Scorte consigliate Carne/Pesce : {pred['fabbisogno_carne_pesce_kg']} kg")
print(f"  - Panetti Pizza da preparare: {pred['panetti_pizza_stimati']}")
```
